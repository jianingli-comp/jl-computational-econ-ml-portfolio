```python
# generate BA graphs
import networkx as nx
import pandas as pd
import os

datasets = [
    (10000, 3, 'ba_10k_edges.csv'),
    (30000, 3, 'ba_30k_edges.csv'),
    (50000, 3, 'ba_50k_edges.csv')
]

for n, m, filename in datasets:
    # Creating BA graphs
    G = nx.barabasi_albert_graph(n, m)
    
    # Save the graphs
    edges = list(G.edges())
    pd.DataFrame(edges, columns=['src', 'dst']).to_csv(filename, index=False)
    
    # Validating
    if os.path.exists(filename):
        df = pd.read_csv(filename)
        if len(df) == len(edges):
            print(f"Success {filename}: {len(edges)} edges")
        else:
            print(f"Fail {filename}: Error - expected {len(edges)}, got {len(df)}")
    else:
        print(f"Fail {filename}: File not created")
```
